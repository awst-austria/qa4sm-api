````python
import pandas as pd
from pathlib import Path

def print_tree(path, prefix="", max_shown=10):
    entries = sorted(Path(path).iterdir(), key=lambda e: (e.is_file(), e.name))
    shown = entries[:max_shown]
    hidden = len(entries) - max_shown

    for i, entry in enumerate(shown):
        is_last = i == len(shown) - 1 and hidden <= 0
        connector = "└── " if is_last else "├── "
        print(prefix + connector + entry.name)
        if entry.is_dir():
            extension = "    " if is_last else "│   "
            print_tree(entry, prefix + extension, max_shown)

    if hidden > 0:
        print(prefix + f"└── ... ({hidden} more)")
````

# Prerequisites

- A token must be created for your account. You can do this at https://qa4sm.eu/ui/user-profile

# Example 1 - Login & save the token
The qa4sm_api package needs your QA4SM token stored in a config file under ~/.qa4smapirc.
You can either set up this file manually as below or use the `setup` function to generate/update
the file.

Example ~/.qa4smapirc
```
[qa4sm.eu]
token: <TOKEN>
username: <USERNAME>
```

If you don't want to create the file yourself, the following command will prompt you for your username and password and store the token as described above.

In your console:
```
qa4sm api setup --help
```

... or in python:

````python

from qa4sm_api.cli import setup_api

setup_api()
````

Now we have a token stored that will be used for subsequent API calls. We can check via the console command



````python
!! qa4sm api check
````

Future API calls will now use token from the .qa4smapirc file so we don't have to enter our credentials every time. Now we set up a new connection in python to send API calls.

````python
from qa4sm_api.client_api import Connection
qa4sm = Connection()
````

# Example 2 - Discover datasets in the service
Retrieve the latest list of available datasets in the service from the server. Datasets that belong to **my user** will also be listed.

````python
table = qa4sm.datasets()
display(table)
````

Using the API we can also get more information for a specific dataset, such as the version.

````python
pd.DataFrame(qa4sm.dataset_info(qa4sm.dataset_id("SMOS_L2")))
````

# Example 4 - Access exisiting validation run

Each validation run has a UUID (see URL). Via the API, access various features of a run.

https://qa4sm.eu/ui/validation-result/9aeb663b-e24e-4541-8331-6ec3e0318d1f

````python
run_id = "9aeb663b-e24e-4541-8331-6ec3e0318d1f" 
qa4sm.validation_status(run_id)    # return status of run (DONE, ERROR, SCHEDULED, CANCELLED, RUNNING)
````

Find out when a validation run was processed and how long it took.

````python
print(qa4sm.validation_time(run_id), "\n")   # list when a run was processing
print(qa4sm.validation_duration(run_id))  # list how long processing took
````

We can download the validation results. Some of these are needed to create the report.

````python
qa4sm.download_results(run_id, "/tmp/example_results")
print_tree("/tmp/example_results")
````

Retrieve the configuration (download as a json file)

````python
config = qa4sm.download_configuration(run_id)
display(config.data)
````

# Example 5 - Submit a new validation run

Based on the config file we downloaded, we submit a new run with similar settings.
- The spatial extent is now smaller
- The validation period is also changed

````python
config['min_lat'] = 40
config['max_lat'] = 50
config['min_lon'] = -10
config['max_lon'] = 10

config['interval_from'] = "2024-10-01"
config['interval_to'] = "2024-10-15"

response = qa4sm.run_validation(config)
````

We can again check the status of the triggered validation run

````python
qa4sm.validation_status(response.name) 
````

The URL is

````python
print(f"https://qa4sm.eu/ui/validation-result/{response.pk}")
````
